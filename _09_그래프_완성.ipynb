### **[실습문제] networkx로 그래프 그리기**
1. 무향 그래프(Undirected Graph)
2. 유향 그래프(Directed Graph)
3. 가중치 그래프(Weight Graph)


```python
import networkx as nx
import matplotlib.pyplot as plt

def draw_edge_graph(edges, nodes=None, pos=None, directed=False, weighted=False, title='Graph', figsize=(4, 3)):
    """
    다양한 형태의 그래프를 그리는 함수
    :param edges: 엣지 리스트
    :param nodes: 노드 리스트/셋
    :param pos: 노드 위치 딕셔너리 (없으면 spring_layout 사용)
    :param directed: 방향 그래프 여부
    :param weighted: 가중치 포함 여부
    :param title: 그래프 제목
    :param figsize: 그림 크기
    """
    # 방향성 설정
    G = nx.DiGraph() if directed else nx.Graph()

    # 노드, 엣지 추가
    if nodes:
        G.add_nodes_from(nodes)
    if weighted:
        G.add_weighted_edges_from(edges)
    else:
        G.add_edges_from(edges)

    # 노드 위치 설정
    if not pos:
        pos = nx.spring_layout(G, seed=42)

    # 시각화
    plt.figure(figsize=figsize)
    nx.draw(
        G, pos, with_labels=True,
        node_color='skyblue', node_size=600,
        edge_color='tab:blue', width=2.5,
        font_size=14, alpha=0.9, arrows=directed
    )

    # 가중치 표시
    if weighted:
        edge_labels = nx.get_edge_attributes(G, 'weight')
        nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=12)

    plt.title(title, fontsize=15)
    plt.axis('off')
    plt.show()


# 노드와 위치 고정
V = {'A', 'B', 'C', 'D'}
pos = {'A': (0, 0), 'B': (0, -1), 'C': (1, -1), 'D': (1, 0)}

# 1. 무방향 그래프
E1 = {('A', 'B'), ('A', 'C'), ('A', 'D'), ('B', 'C'), ('C', 'D')} # 집합 or 리스트로 표현 가능
draw_edge_graph(E1, V, pos, directed=False, weighted=False, title='Undirected Graph')

# 2. 방향 그래프
E2 = [('A', 'B'), ('A', 'C'), ('B', 'C'), ('C', 'D'), ('D', 'A')]
draw_edge_graph(E2, V, pos, directed=True, weighted=False, title='Directed Graph')

# 3. 가중치 무방향 그래프
E3 = [('A', 'B', 4), ('A', 'C', 1), ('A', 'D', 2), ('B', 'C', 5), ('C', 'D', 3)]
draw_edge_graph(E3, V, pos, directed=False, weighted=True, title='Weighted Undirected Graph')

# 4. 가중치 방향 그래프
draw_edge_graph(E3, V, pos, directed=True, weighted=True, title='Weighted Directed Graph')
```

### **[참고] PageRank 알고리즘**
- d = 0.85             # damping factor
- T = 8                # 반복 횟수(수업용으로 8회)

- **두 값을 변경하고 그래프가 어떻게 변하는지 확인한다.**

```python
# ==========================================
# PageRank: 초기 균등 -> 반복 적용 -> 수렴 결과 시각화 (Colab OK)
# ==========================================

# 1) 라이브러리 불러오기
import networkx as nx           # 그래프 구조와 알고리즘
import matplotlib.pyplot as plt # 시각화
import numpy as np              # 수치 계산
import pandas as pd             # 표/결과 정리

# 2) 예제 그래프 정의 (방향 그래프)
#   A -> B, C
#   B -> C, D
#   C -> A
#   D -> C, E
#   E -> C
edges = [
    ("A", "B"), ("A", "C"),
    ("B", "C"), ("B", "D"),
    ("C", "A"),
    ("D", "C"), ("D", "E"),
    ("E", "C")
]
G = nx.DiGraph()
G.add_edges_from(edges)


# 3) 공통 파라미터
d = 0.85             # damping factor
T = 8                # 반복 횟수(수업용으로 8회)
nodes = list(G.nodes())
N = len(nodes)
seed = 42            # 레이아웃 고정용 시드
pos = nx.spring_layout(G, seed=seed)



# ---------- 유틸: 전이행렬 M 만들기(열합=1), dangling 처리 ----------
def build_column_stochastic_M(G, nodes):
    """열이 from, 행이 to. 각 열 합이 1이 되도록 구성"""
    idx = {n: i for i, n in enumerate(nodes)}
    N = len(nodes)
    M = np.zeros((N, N), dtype=float)
    out_degs = dict(G.out_degree())
    for j in nodes:
        outdeg = out_degs[j]
        if outdeg > 0:
            for _, i in G.out_edges(j):
                M[idx[i], idx[j]] = 1.0 / outdeg
        # outdeg==0인 경우는 여기서 0으로 두고, 반복식에서 dangling으로 별도 처리
    return M


def dangling_vector(G, nodes):
    """out_degree==0인 노드에 1, 아니면 0인 벡터"""
    out_degs = dict(G.out_degree())
    dvec = np.array([1.0 if out_degs[n] == 0 else 0.0 for n in nodes], dtype=float)
    return dvec


# ---------- 핵심: PageRank 반복 함수 ----------
def pagerank_iterative(G, d=0.85, T=8, init=None, nodes=None):
    """
    PR^{t+1} = (1-d)/N * 1 + d * ( M @ PR^{t} + dangling_mass * (1/N) )
    - M: column-stochastic(열합=1) 전이행렬
    - dangling 노드(출링크 0)의 질량은 균등 분배
    """
    if nodes is None:
        nodes = list(G.nodes())
    N = len(nodes)

    # 초기 PR (균등)
    if init is None:
        pr = np.ones(N, dtype=float) / N
    else:
        pr = np.array(init, dtype=float)
        pr = pr / pr.sum()  # 안전 차원 정규화

    # 전이행렬과 dangling 벡터 구성
    M = build_column_stochastic_M(G, nodes)
    dvec = dangling_vector(G, nodes)

    history = [pr.copy()]
    for _ in range(T):
        base = (1 - d) / N
        dangling_mass = (dvec @ pr)  # dangling 노드가 보유한 질량
        pr = base + d * (M @ pr) + d * (dangling_mass / N)
        history.append(pr.copy())

    return np.array(history), M


# ---------- 시각화 유틸 ----------
def draw_graph_with_weights(G, pos, weights, title, cmap=False):
    """
    주어진 PR 가중치에 따라 노드 크기/라벨을 그린다.
    - weights: 노드 순서(nodes)의 값 배열(또는 dict)
    """
    if isinstance(weights, dict):
        w_arr = np.array([weights[n] for n in G.nodes()])
    else:
        w_arr = np.array(weights)

    # 노드 크기 스케일 (보기 좋게 배율 적용)
    sizes = 3000 * (w_arr / w_arr.sum())

    plt.figure(figsize=(7, 5))
    nx.draw_networkx_edges(G, pos, arrows=True, arrowstyle="->", width=1.6, alpha=0.7)
    nx.draw_networkx_nodes(G, pos, node_size=sizes, edgecolors="k", alpha=0.6)
    nx.draw_networkx_labels(G, pos, font_size=12)

    # 노드 아래에 PR 값 주석
    for n, (x, y) in pos.items():
        plt.text(x, y - 0.12, f"{w_arr[list(G.nodes()).index(n)]:.4f}", ha="center", fontsize=9, color="red")

    plt.title(title)
    plt.axis("off")
    plt.show()




# ---------- 4) 초기 상태(균등 PR) 시각화 ----------
init_pr = np.ones(N) / N
draw_graph_with_weights(G, pos, init_pr, "초기 상태: 균등 PageRank (모두 동일 크기)")

# ---------- 5) 반복 수행 ----------
history, M = pagerank_iterative(G, d=d, T=T, nodes=nodes)
df_hist = pd.DataFrame(history, columns=nodes)
df_hist.index.name = "iteration"

print("=== 반복 갱신 결과(처음~마지막) ===")
display(df_hist.round(6))

# ---------- 6) 반복 후(수렴 근처) 상태 시각화 ----------
final_pr = history[-1]
draw_graph_with_weights(G, pos, final_pr, f"반복 후 상태: PageRank (T={T})")

# ---------- 7) 반복 과정 수렴 곡선 시각화 ----------
plt.figure(figsize=(7, 5))
for n in nodes:
    plt.plot(df_hist.index, df_hist[n], marker="o", label=n)
plt.xlabel("iteration")
plt.ylabel("PageRank")
plt.title("PageRank 반복 갱신 수렴 과정")
plt.grid(True)
plt.legend()
plt.show()


# ---------- 8) NetworkX 내장 결과와 비교 ----------
pr_nx = nx.pagerank(G, alpha=d, max_iter=100, tol=1e-12)
df_compare = pd.DataFrame({
    "node": nodes,
    "nx_pagerank": [pr_nx[n] for n in nodes],
    "iter_last":   [final_pr[list(G.nodes()).index(n)] for n in nodes]
})
df_compare["abs_diff"] = (df_compare["nx_pagerank"] - df_compare["iter_last"]).abs()
print("=== NetworkX vs 수식 반복 결과 비교 ===")
display(df_compare.round(6).sort_values("node").reset_index(drop=True))

# # ---------- 9) 중간 단계 스냅샷(선택 반복차수) ----------
# snap_iters = [0, 1, 2, 4, T]  # 필요시 조정
# for k in snap_iters:
#     draw_graph_with_weights(G, pos, history[k], f"스냅샷: iteration={k}")
```

## **1-2. 그래프 표현**

그래프에서 **간선의 집합을 표현**하는 방법
1. **엣지 리스트(Edge List)**
    - 그래프의 연결 관계를 간선(시작노드, 끝노드) or (시작노드, 끝노드, 가중치) 로 나타내는 방법
2. **인접 행렬(Adjacency Matrix)**
    - 그래프의 연결 관계를 2차원 배열로 나타내는 방법
3. **인접 리스트(Adjacency List)**\
    - 그래프의 연결 관계를 연결 리스트로 나타내는 방법

### **1) 엣지 리스트(Edge List)로 그래프 표현하기**

```python
# 무향 그래프
edges_list1 = [
    ('U', 'V'), ('V', 'U'),
    ('U', 'W'), ('W', 'U'),
    ('W', 'Y'), ('Y', 'W'),
    ('V', 'W'), ('W', 'V'),
    ('V', 'X'), ('X', 'V')
]

# 유향 그래프
edges_list2 = [
    ('U', 'V'),
    ('V', 'U'),
    ('V', 'W')
]
print(f"✅ 엣지_리스트(undirected):\n{edges_list1}")
print(f"✅ 엣지_리스트(directed):\n{edges_list2}")

# 그래프 시각화
draw_edge_graph(edges_list1, None, None, directed=False, weighted=False, title='UnDirected Graph')
draw_edge_graph(edges_list2, None, None, directed=True, weighted=False, title='Directed Graph')
```

### **2) 인접 행렬(Adjacency Matrix)로 그래프 표현하기**

```python
import numpy as np

def create_adj_matrix(edges, directed=False):
    """엣지 리스트로부터 인접 행렬 생성"""
    nodes = sorted(set(node for edge in edges for node in edge))
    node_to_index = {node: idx for idx, node in enumerate(nodes)}
    num_nodes = len(nodes)

    adj_matrix = np.zeros((num_nodes, num_nodes), dtype=int)

    for start_node, end_node in edges:
        i, j = node_to_index[start_node], node_to_index[end_node]
        adj_matrix[i, j] = 1
        if not directed: # 무향 그래프는 대칭을 적용한다.
            adj_matrix[j, i] = 1

    return adj_matrix


# 엣지 리스트 정의
edges_list1 = [
    ('U', 'V'),
    ('U', 'W'),
    ('W', 'Y'),
    ('V', 'W'),
    ('V', 'X')
]

# 인접 행렬 생성
adj_matrix_a = create_adj_matrix(edges_list1, directed=False)
print(f"✅ 인접_행렬(undirected):\n{adj_matrix_a}")
adj_matrix_b = create_adj_matrix(edges_list2, directed=True)
print(f"✅ 인접_행렬(directed):\n{adj_matrix_b}")
```

### **3) 인접 리스트(Adjacency List)로 그래프 표현하기**

```python
from collections import defaultdict

def create_adj_list(edges, directed=False):
    """엣지 리스트로부터 인접 리스트를 생성하는 함수"""
    adj_list = defaultdict(list)

    for start, end in edges:
        adj_list[start].append(end)
        if not directed:
            adj_list[end].append(start)

    return dict(adj_list)  # 필요 시 defaultdict 유지 가능


# 인접 리스트 생성
adj_list_a = create_adj_list(edges_list1, directed=False)
print(f"✅ 인접_리스트(undirected):\n{adj_list_a}")
adj_list_b = create_adj_list(edges_list2, directed=True)
print(f"✅ 인접_리스트(directed):\n{adj_list_b}")
```

## **1-3. 그래프 탐색**

### **1) 깊이 우선 탐색(Depth-First Search, DFS)**

- 인접 리스트를 이용하여 그래프 탐색

```python
# 인접 리스트를 이용하는방법
from collections import defaultdict

# 그래프 정의 (무방향 그래프)
graph = defaultdict(list)

# 엣지 리스트 정의
edges_list1 = [
    ('U', 'V'),
    ('U', 'W'),
    ('W', 'Y'),
    ('V', 'W'),
    ('V', 'X')
]

# 그래프를 인접 리스트로 생성
graph = create_adj_list(edges_list1, directed=False)
# 이웃 노드를 정렬: V -> W -> X 순으로 탐색되도록
for node in graph:
    graph[node] = sorted(graph[node], reverse=True)
print(f"✅ 인접_리스트_그래프 (undirected):\n{graph}")


# DFS 함수 (반복문 방식)
def dfs_stack_iter_list(start):
    visited = set()
    stack = [start]

    print(f"\nDFS 시작 (start: {start})")
    while stack:
        current = stack.pop()
        if current not in visited:
            print(f"방문: {current}")
            visited.add(current)
            # 스택에는 나중에 방문할 노드가 위로 가게 push
            for neighbor in graph[current]:
                if neighbor not in visited:
                    stack.append(neighbor)
    return visited


# DFS 함수 (재귀 방식)
def dfs_stack_recur_list(node, visited=None):
    if visited is None:
        visited = set()
        print(f"\nDFS 시작 (start: {node})")

    print(f"방문: {node}")
    visited.add(node)

    for neighbor in graph[node]:
        if neighbor not in visited:
            dfs_stack_recur_list(neighbor, visited)

    return visited


# 탐색 실행
visited_order = dfs_stack_iter_list('U')
print("✅ DFS 탐색 결과(반복문): ", visited_order)
visited_order = dfs_stack_recur_list('U')
print("✅ DFS 탐색 결과(재귀문): ", visited_order)
```

- 인접 행렬 이용하여 그래프 탐색

```python
import numpy as np

# 노드와 인덱스 매핑
nodes = ['U', 'V', 'W', 'X', 'Y']
node_to_idx = {node: idx for idx, node in enumerate(nodes)}
idx_to_node = {idx: node for node, idx in node_to_idx.items()}
N = len(nodes)

# 엣지 리스트 정의 (무방향)
edges_list1 = [
    ('U', 'V'),
    ('U', 'W'),
    ('W', 'Y'),
    ('V', 'W'),
    ('V', 'X')
]
# 인접 행렬 생성
adj_matrix = create_adj_matrix(edges_list1, directed=False)
print(f"✅ 인접_행렬(undirected):\n{adj_matrix}")


# DFS 함수 (반복 방식)
def dfs_stack_iter_matrix(start_node):
    visited = set()
    stack = [node_to_idx[start_node]]
    print(f"\n[반복 DFS 시작] 시작 노드: {start_node}")

    while stack:
        current = stack.pop()
        if current not in visited:
            print(f"방문: {idx_to_node[current]}")
            visited.add(current)

            neighbors = [i for i in range(N) if adj_matrix[current][i] == 1 and i not in visited]
            for neighbor in sorted(neighbors, reverse=True):
                stack.append(neighbor)

    return {idx_to_node[i] for i in visited}


# DFS 함수 (재귀 방식)
def dfs_stack_recur_matrix(current_idx, visited=None):
    if visited is None:
        visited = set()
        print(f"\n[재귀 DFS 시작] 시작 노드: {idx_to_node[current_idx]}")

    # 방문하지 않은 경우에만 처리
    if current_idx not in visited:
        print(f"방문: {idx_to_node[current_idx]}")
        visited.add(current_idx)

        neighbors = [i for i in range(N) if adj_matrix[current_idx][i] == 1]
        for neighbor in sorted(neighbors, reverse=True):
            dfs_stack_recur_matrix(neighbor, visited)

    return {idx_to_node[i] for i in visited}


# ▣ 실행
visited_iter = dfs_stack_iter_matrix('U')
print("✅ DFS 탐색 결과(반복문): ", visited_iter)

visited_recur = dfs_stack_recur_matrix(node_to_idx['U'])
print("✅ DFS 탐색 결과(재귀문): ", visited_recur)
```

### **2) 너비 우선 탐색(Breadth-First Search, BFS)**

- 인접 리스트를 이용하여 그래프 탐색
    - 보통 BFS(너비 우선 탐색)는 큐(Queue) 를 사용하는 반복문 방식으로 구현

```python
# 인접 리스트를 이용하는방법
from collections import defaultdict, deque

# 그래프 정의 (무방향 그래프)
graph = defaultdict(list)

# 엣지 리스트 정의
edges_list1 = [
    ('U', 'V'),
    ('U', 'W'),
    ('W', 'Y'),
    ('V', 'W'),
    ('V', 'X')
]

# 그래프를 인접 리스트로 생성
graph = create_adj_list(edges_list1, directed=False)
# 이웃 노드를 정렬: V -> W -> X 순으로 탐색되도록
for node in graph:
    graph[node] = sorted(graph[node], reverse=True)
print(f"✅ 인접_리스트_그래프 (undirected):\n{graph}")


# BFS 함수 (반복문 방식)
def bfs_queue_iter_list(start):
    visited = set()
    queue = deque([start])

    print(f"\nBFS 시작 (start: {start})")
    while queue:
        current = queue.popleft()
        if current not in visited:
            print(f"방문: {current}")
            visited.add(current)

            for neighbor in graph[current]:
                if neighbor not in visited:
                    queue.append(neighbor)
    return visited


# 탐색 실행
visited_order = bfs_queue_iter_list('U')
print("✅ BFS 탐색 결과(반복문): ", visited_order)
```

## **1-4. 최소 신장 트리(MST, Minimum Spanning Tree)**

### **1) 스패닝 트리(Spanning tree) :  신장 트리**

- 그래프 내 모든 정점을 포함하는 트리
- 정점 n은 n-1개의 간선을 갖는다
- 모든 정점들이 연결 되어 있어야 하고
- 사이클을 포함해서는 안된다.

```python
# DFS를 이용한 spanning tree(인접행렬 방식)
def st_dfs(vtx, adj, s, visited) :
    visited[s] = True               # 현재 정점 s를 visited에 추가함
    for v in range(len(vtx)) :      # 인접행렬
        if adj[s][v] != 0 :         # 모든 간선 (s,v)에 대해
            if visited[v]==False:   # v를 아직 방문하지 않았으면
                print("(", vtx[s], vtx[v], ")", end=' ')  # 간선 출력
                st_dfs(vtx, adj, v, visited)


vtx = ['U','V','W','X','Y']
edges_list = [
    ('U', 'V'),
    ('U', 'W'),
    ('V', 'W'),
    ('V', 'X'),
    ('W', 'X'),
    ('W', 'Y'),
    ('X', 'Y')
]
# 인접 행렬 생성
adj_matrix = create_adj_matrix(edges_list, directed=False)
print(f"✅ 인접_행렬(undirected):\n{adj_matrix}")

# spanning tree
st_dfs(vtx, adj_matrix, 0, [False]*len(vtx))
print()
```

### **2) 최소 스패닝 트리 (Minimum Spanning Tree)**
- 가중치 그래프의 여러 신장 트리 중에서 간선의 가중치 합이 최소인 트리
컴퓨터 인터넷 네트워크와 자동차 내비게이션 등에 많이 사용된다
- Kruskal 알고리즘 & Prim 알고리즘

### **@프림(Prim) 알고리즘**
한 정점에서 시작하여 가장 짧은 연결 간선을 반복적으로 선택

```python
edges_weight_list = [
    ('A', 'B', 25),
    ('A', 'D', 12),
    ('B', 'C', 10),
    ('B', 'E', 15),
    ('C', 'G', 16),
    ('E', 'G', 14),
    ('E', 'D', 17),
    ('E', 'F', 19),
    ('D', 'F', 37),
    ('F', 'G', 42)
]
```

- 강의자료 방법: table 이용
    - selected & dist 사용

```python
import sys

# ▣ 노드 인덱스 매핑
nodes = ['A', 'B', 'C', 'D', 'E', 'F', 'G']
idx = {node: i for i, node in enumerate(nodes)}
ridx = {i: node for node, i in idx.items()}
N = len(nodes)

# ▣ 인접 행렬 초기화 (무한대)
INF = float('inf')
adj_matrix = [[INF] * N for _ in range(N)]

# ▣ 인접 행렬 채우기 (무방향)
for u, v, w in edges_weight_list:
    i, j = idx[u], idx[v]
    adj_matrix[i][j] = w
    adj_matrix[j][i] = w

print(f"✅ 인접_행렬(undirected):\n{adj_matrix}")


# ▣ Prim 알고리즘 (배열 기반)
def prim_with_table(start_node):
    start = idx[start_node]
    selected = [False] * N
    dist = [INF] * N
    parent = [None] * N

    dist[start] = 0  # 시작점 가중치 0

    print(f"✅ Prim 알고리즘 시작 (Start Node: {start_node},[{start}])\n")

    for step in range(N):
        # 최소 가중치 정점 선택
        min_val = INF
        u = -1
        for v in range(N):
            if not selected[v] and dist[v] < min_val:
                min_val = dist[v]
                u = v

        selected[u] = True
        print(f"▶ Step {step + 1}: 선택된 노드: {ridx[u]}")
        print("   selected =", ['T' if x else 'F' for x in selected])
        print("   dist     =", [dist[i] if dist[i] != INF else '∞' for i in range(N)])
        print()

        # dist 배열 업데이트
        for v in range(N):
            if adj_matrix[u][v] != INF and not selected[v] and adj_matrix[u][v] < dist[v]:
                dist[v] = adj_matrix[u][v]
                parent[v] = u

    # MST 출력
    total_cost = sum([adj_matrix[i][parent[i]] for i in range(N) if parent[i] is not None])
    print("✅ 최종 MST 간선들:")
    for v in range(N):
        if parent[v] is not None:
            print(f"{ridx[parent[v]]} - {ridx[v]} (가중치: {adj_matrix[v][parent[v]]})")
    print(f"총 비용: {total_cost}")

# 탐색 실행
prim_with_table('A')
```

- 연결 리스트 이용 방법
    - heapq 모듈 : 최소 힙(min-heap) 자료구조를 지원하는 표준 라이브러리

```python
import heapq
from collections import defaultdict

# 그래프를 인접 리스트로 생성 (무방향 그래프)
graph = defaultdict(list)
for u, v, w in edges_weight_list:
    graph[u].append((w, v))
    graph[v].append((w, u))
print(f"✅ 인접_리스트_그래프 (undirected):\n{graph}")


# Prim 알고리즘 단계별 시뮬레이션
def prim_mst_step_by_step(start):
    visited = set()
    min_heap = [(0, start, None)]  # (가중치, 현재노드, 이전노드)
    total_weight = 0
    step = 0

    print(f"\n✅ Prim 알고리즘 시작 (Start: {start})\n")

    while min_heap:
        weight, current, prev = heapq.heappop(min_heap)
        if current in visited:
            continue

        visited.add(current)

        if prev:
            total_weight += weight
            step += 1
            print(f"▶ Step {step}: 선택된 간선: {prev} — {current} (가중치: {weight})")
            print(f"    현재 누적 가중치: {total_weight}")
            print(f"    방문한 노드들: {sorted(visited)}\n")

        for edge_weight, neighbor in graph[current]:
            if neighbor not in visited:
                heapq.heappush(min_heap, (edge_weight, neighbor, current))

    print(f"✅ MST 완성! 총 가중치: {total_weight}")

# 탐색 실행
prim_mst_step_by_step('A')
```

- 그래프 시각화

```python
# 그래프 시각화
import networkx as nx
import matplotlib.pyplot as plt

# MST 그래프 그리기
def draw_mst_graph(edges, pos=False, figsize=(4, 3)):
    # 그래프 생성 및 엣지 추가
    G = nx.Graph()
    G.add_weighted_edges_from(edges)

    # 최소 스패닝 트리(MST) 찾기
    mst = nx.minimum_spanning_tree(G)

    #---------------------
    # MST 가중치 합계 계산
    print( mst.edges(data=True) )
    total_weight = sum(data['weight'] for u, v, data in mst.edges(data=True))
    print("MST 전체 가중치 합:", total_weight)
     #--------------------

    # 원본 그래프와 MST의 엣지를 구분하기 위한 스타일 설정
    if not pos:
        pos = nx.spring_layout(G)  # 노드 위치 자동 배치

    # 시각화 시작
    plt.figure(figsize=figsize)
    nx.draw(G, pos, with_labels=True, node_color='lightblue', node_size=700, font_size=12)
    nx.draw_networkx_edge_labels(G, pos, edge_labels={(u, v): d['weight'] for u, v, d in G.edges(data=True)})

    # MST의 엣지를 강조하여 그리기
    nx.draw_networkx_edges(mst, pos, edge_color='red', width=2)
    plt.title("Minimum Spanning Tree (MST) Visualization")
    plt.show()


pos = {'A':(-1,1),
       'B':(0,2),
       'C':(1,1),
       'D':(-2,-1),
       'E':(0,0),
       'F':(0,-2),
       'G':(2,-1),
      }

draw_mst_graph(edges_weight_list, pos)
```

### [참고] **@크루스칼(Kruskal) 알고리즘**
간선을 가중치 오름차순 정렬, Union-Find 구조로 사이클 방지

```python
# ▣ Union-Find 자료구조 : 사이클을 방지하거나 연결 여부를 빠르게 판단하기 위해 사용하는 대표적인 자료구조
'''
- parent: 각 원소가 속한 집합의 대표 원소(루트)를 저장
- find(x): 원소 x가 속한 집합의 대표(루트)를 찾음
- union(x, y): 두 원소 x, y가 속한 집합을 합침
'''
parent = {}

def find(x):
    if parent[x] != x:
        parent[x] = find(parent[x])
    return parent[x]

def union(x, y):
    root_x = find(x)
    root_y = find(y)
    if root_x != root_y:
        parent[root_y] = root_x
        return True
    return False


# Kruskal 알고리즘
def kruskal_mst(edges):
    mst = []
    total_weight = 0
    nodes = set(sum(([u, v] for u, v, _ in edges), [])) # 리스트 내의 리스트들을 평탄화(flatten)해서 set 타입으로
    print(f'nodes:{nodes}')

    # 초기화
    for node in nodes:
        parent[node] = node
    print(f'parent:{parent}')

    # 간선 가중치 오름차순 정렬
    edges_sorted = sorted(edges, key=lambda x: x[2])
    print(f'edges_sorted:{edges_sorted}')

    for u, v, weight in edges_sorted:
        if union(u, v):                 # u와 v가 서로 다른 집합(트리)에 속하면 합치고 간선을 선택한다, 그렇지 않으면 (사이클 생기므로) 무시한다
            mst.append((u, v, weight))
            total_weight += weight

    return mst, total_weight


# Kruskal 실행
mst_kruskal, total_kruskal = kruskal_mst(edges_weight_list)
print("✅ Kruskal MST 결과:")
for u, v, w in mst_kruskal:
    print(f"{u} - {v}: {w}")
print(f"✅ 총 가중치: {total_kruskal}")
```

### **[실습문제] 가중치가 있는 그래프용 데이터를 이용하여 MST를 찾고 시각화 하시오.**
- 데이터 (source, target, weight) 구조


- 방법1: 배포된 social_network_weighted.csv 데이터를 이용하여 Prim 알고리즘을 이용하여 MST를 찾고, 시각화한다.

```python
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

# CSV 파일 읽기
csv_file = 'social_network_weighted.csv'  # 실제 파일 경로로 변경하세요
df = pd.read_csv(csv_file)
display(df.head())

# 그래프 객체 생성
G = nx.from_pandas_edgelist(df, 'source', 'target', edge_attr='weight')
print(f"✅ 총 엣지수: {len(G.edges)}")
print(f"✅ 총 노드수: {len(G.nodes)}\n{G.nodes}")

# 가중치 엣지 목록
edges_weight_list = [(u, v, d['weight']) for u, v, d in G.edges(data=True)]
print(f"✅ 가중치 엣지 리스트: {edges_weight_list}")


# MST 그래프 그리기
def draw_mst_graph(edges, pos=False, figsize=(4, 3)):
    # 그래프 생성 및 엣지 추가
    G = nx.Graph()
    G.add_weighted_edges_from(edges)

    # 최소 스패닝 트리(MST) 찾기
    mst = nx.minimum_spanning_tree(G)

    #---------------------
    # MST 가중치 합계 계산
    print( mst.edges(data=True) )
    total_weight = sum(data['weight'] for u, v, data in mst.edges(data=True))
    print("MST 전체 가중치 합:", total_weight)
     #--------------------

    # 원본 그래프와 MST의 엣지를 구분하기 위한 스타일 설정
    if not pos:
        pos = nx.spring_layout(G)  # 노드 위치 자동 배치

    # 시각화 시작
    plt.figure(figsize=(10, 10))
    nx.draw(G, pos, with_labels=True, node_color='lightblue', node_size=400, font_size=10)
    nx.draw_networkx_edge_labels(G, pos, edge_labels={(u, v): d['weight'] for u, v, d in G.edges(data=True)})

    # MST의 엣지를 강조하여 그리기
    nx.draw_networkx_edges(mst, pos, edge_color='red', width=2)
    plt.title("Minimum Spanning Tree (MST) Visualization")
    plt.show()

# Prim 알고리즘으로 MST 탐색 실행
prim_mst_step_by_step('Alice')
draw_mst_graph(edges_weight_list)
```

# [참고] 그래프 응용

### **SNS 관계 표현 (파일 읽어서)**
- source, target, weight

```python
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

# CSV 파일 읽기
csv_file = 'social_network_weighted.csv'  # 실제 파일 경로로 변경하세요
df = pd.read_csv(csv_file)
display(df)

# 그래프 객체 생성
G = nx.from_pandas_edgelist(df, 'source', 'target', edge_attr='weight')
print(f"✅ 총 엣지수: {len(G.edges)}")
print(f"✅ 총 노드수: {len(G.nodes)}\n{G.nodes}")

# 노드 위치 설정 (spring layout 사용)
pos = nx.spring_layout(G, k=0.3, seed=42)
#-----------------------------------

# 노드 그리기
plt.figure(figsize=(10, 10))
nx.draw_networkx_nodes(G, pos, node_size=100, node_color='skyblue')
nx.draw_networkx_labels(G, pos, font_size=9)  # 노드 라벨 그리기

# 엣지 그리기 (가중치 기반 굵기)
# weights = [d['weight'] * 0.3 for (u, v, d) in G.edges(data=True)]
edge_labels = nx.get_edge_attributes(G, 'weight')
nx.draw_networkx_edges(G, pos, width=0.5, alpha=0.6)
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=7) # 엣지 라벨 (가중치)

plt.title("소셜 네트워크 그래프", fontsize=15)
plt.axis('off')
plt.show()
```

- 가중치에 따른 노드와 엣지의 시각화


| 요소          | 단계  | 기준                           | 시각적 요소                  |
| ----------- | --- | ---------------------------- | ----------------------- |
| **엣지 굵기/색** | 5단계 | `weight` 값 범위에 따라            | 굵기 (선 두께) + 색상 (진하기 변화) |
| **노드 크기/색** | 3단계 | 연결된 가중치 총합 (`node strength`) | 크기 (원 크기) + 색상 (밝기 차이)  |


```python
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors

# CSV 파일 읽기
csv_file = 'social_network_weighted.csv'  # 실제 파일 경로로 변경하세요
df = pd.read_csv(csv_file)

# 그래프 객체 생성
G = nx.from_pandas_edgelist(df, 'source', 'target', edge_attr='weight')

# 노드 위치 설정 (spring layout 사용)
pos = nx.spring_layout(G, k=0.3, seed=42)
#-----------------------------------


# 노드 strength 계산 (가중치 합)
node_strength = {}
for node in G.nodes():
    node_strength[node] = sum(d['weight'] for u, v, d in G.edges(node, data=True))

# 노드 크기와 색상 (3단계)
node_strength_values = list(node_strength.values())
min_s, max_s = min(node_strength_values), max(node_strength_values)
node_colors = []
node_sizes = []
for n in G.nodes():
    strength = node_strength[n]
    if strength < min_s + (max_s - min_s) / 3:
        node_colors.append('lightblue')
        node_sizes.append(100)
    elif strength < min_s + 2 * (max_s - min_s) / 3:
        node_colors.append('deepskyblue')
        node_sizes.append(300)
    else:
        node_colors.append('blue')
        node_sizes.append(600)

# ➖ 엣지 색상과 굵기 (5단계)
edge_colors = []
edge_widths = []
for u, v, d in G.edges(data=True):
    w = d['weight']
    if w <= 2:
        edge_colors.append('#cccccc')
        edge_widths.append(0.5)
    elif w <= 4:
        edge_colors.append('#999999')
        edge_widths.append(1)
    elif w <= 6:
        edge_colors.append('#666666')
        edge_widths.append(2)
    elif w <= 8:
        edge_colors.append('#333333')
        edge_widths.append(3)
    else:
        edge_colors.append('#000000')
        edge_widths.append(4)



# 노드 그리기 : 가중치 적용 (크기,색상:node_sizes,node_colors)
plt.figure(figsize=(10, 10))
nx.draw_networkx_nodes(G, pos, node_size=node_sizes, node_color=node_colors, alpha=0.9)
nx.draw_networkx_labels(G, pos, font_size=9)

# 엣지 그리기 : 가중치 적용 (굵기,색상:edge_widths,edge_colors)
edge_labels = nx.get_edge_attributes(G, 'weight')
nx.draw_networkx_edges(G, pos, width=edge_widths, edge_color=edge_colors, alpha=0.6)
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=7)

plt.title("소셜 네트워크 그래프(가중치 적용)", fontsize=15)
plt.axis('off')
plt.show()
```

### **Tarjan’s algorithm 적용** :  **SCC 찾기**
- 방향 그래프에서 **강하게 연결된 컴포넌트(SCC,Strongly Connected Components**)를 선형 시간 내에 찾는 알고리즘
- 이름은 개발자인 Robert Tarjan의 이름에서 왔고, DFS 기반 알고리즘
- SCC란
    - SCC는 노드 집합 내의 모든 노드가 서로 도달 가능한 서브그래프
    - ex: A → B → C → A (서로 돌아옴) ⇒ {A, B, C} 는 하나의 SCC
    

```python
import pandas as pd
import networkx as nx

# CSV 파일 로드
df = pd.read_csv('social_network_weighted.csv')

# 방향 그래프(DiGraph) 생성
# source → target 방향으로 간선 설정
DG = nx.from_pandas_edgelist(df, 'source', 'target', edge_attr='weight', create_using=nx.DiGraph)

# Tarjan 알고리즘 적용: 강한 연결 요소(SCC) 찾기
scc_list = list(nx.strongly_connected_components(DG))

# 출력 결과
print(f"✅ 총 {len(scc_list)}개의 SCC가 발견되었습니다.\n")
for i, scc in enumerate(scc_list, 1):
    print(f"SCC {i}: {scc}")
    print(len(scc))
```
